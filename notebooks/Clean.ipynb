# Prepare the Yelp business data

Put the original Yelp business JSON Lines export at `data/yelp_academic_dataset_business.json`, then run the code cell. The notebook keeps open restaurant listings tagged with Restaurants, normalizes columns and categories, removes duplicate business IDs, and writes `data/cleaned.csv`.

The raw export is not included in this project. When it is absent, this notebook reports the expected location and exits cleanly.

```python
from pathlib import Path
import os
import sys

PROJECT_ROOT = next(parent for parent in (Path.cwd(), *Path.cwd().parents) if (parent / "src").is_dir())
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

from src.data_cleaning import load_yelp_json

raw_path = PROJECT_ROOT / "data" / "yelp_academic_dataset_business.json"
output_path = PROJECT_ROOT / "data" / "cleaned.csv"
if not raw_path.exists():
    print(f"Raw Yelp JSON Lines file not found. Place it here: {raw_path}")
else:
    restaurants, quality = load_yelp_json(raw_path)
    restaurants.to_csv(output_path, index=False)
    print(f"Saved {len(restaurants):,} restaurants to {output_path}")
    print("Data quality:", quality)
```
